# **Exploratory Data Analysis(EDA)**:

```python
#Gender
df.columns
```

```python
ax= sns.countplot(x= 'Gender', data= df)

for bars in ax.containers:
 ax.bar_label(bars)
```

```python
df.groupby(['Gender'], as_index= False)['Amount'].sum()
```

```python
#Total amount vs Gender
Sales_gen = df.groupby(['Gender'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False)
# ascending=False, it tells the sort_values function to arrange your data in descending order
# as_index=False in a groupby operation keeps grouped columns as regular data columns instead of turning them into row index labels.
ax1 = sns.barplot(x= 'Gender', y= 'Amount', data= Sales_gen)
for bars in ax1.containers:
 ax1.bar_label(bars)
```

From above graphs we can conclude that, most of buyers are female, even the purchasing power is greather than male.

```python
#Age
df.columns
```

```python
ax=sns.countplot(data= df, x= 'Age Group', hue='Gender')
sns.set(rc= {'figure.figsize':(10,8)})
for bars in ax.containers:
 ax.bar_label(bars)
```

```python
#Total amount vs age group
Sales_age = df.groupby(['Age Group'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False)
sns.barplot(x= 'Age Group', y= 'Amount', data= Sales_age)
sns.set(rc= {'figure.figsize':(20,10)})
ax2 = sns.barplot(x= 'Age Group', y= 'Amount', data= Sales_age)
for bars in ax2.containers:
 ax2.bar_label(bars)
```

From above graphs we can conclude that, most of buyers are of age group between 26-35 Years.

```python
#from matplotlib import figure
#State #total number of orders from top 10 states
Sales_state = df.groupby(['State'], as_index= False)['Orders'].sum().sort_values(by='Orders', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(15,5)})
sns.barplot(x= 'State', y= 'Orders', data= Sales_state)
ax2 = sns.barplot(x= 'State', y= 'Orders', data= Sales_state)
for bars in ax2.containers:
  ax2.bar_label(bars)
```

```python
#total amount vs sales from top 10 states
Sales_state = df.groupby(['State'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(20,10)})
sns.barplot(x= 'State', y= 'Amount', data= Sales_state)
ax2 = sns.barplot(x= 'State', y= 'Amount', data= Sales_state)
for bars in ax2.containers:
 ax2.bar_label(bars)
```

From above graphs we can conclude that, most of Orders from UP, MH, Karnataka respectively and Sales vs Amount from UP,MH, Karnataka.

```python
#Marital status
ax=sns.countplot(x= 'Married',data= df)
sns.set(rc= {'figure.figsize':(6,4)})
for bars in ax.containers:
 ax.bar_label(bars)
```

```python
Sales_maritalStatus = df.groupby(['Married', 'Gender'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(15,8)})
ax4 = sns.barplot(x= 'Married', y= 'Amount', data= Sales_maritalStatus, hue='Gender')
# ax4 = sns.barplot(x= 'Marital_Status', y= 'Amount', data= Sales_maritalStatus)
for bars in ax4.containers:
 ax4.bar_label(bars)
```

From above graphs we can conclude that, most of buyers are Married Women and they have a more purchasing power than Married Male.

```python
#Occupation
sns.set(rc= {'figure.figsize':(15,8)})
ax=sns.countplot(x= 'Occupation', data= df)
for bars in ax.containers:
 ax.bar_label(bars)
```

```python
Sales_Occupation = df.groupby(['Occupation'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(20,10)})
sns.barplot(x= 'Occupation', y= 'Amount', data= Sales_Occupation)
ax3 = sns.barplot(x= 'Occupation', y= 'Amount', data= Sales_Occupation)
for bars in ax3.containers:
 ax3.bar_label(bars)
```

From above graphs we can conclude that, most of buyers are working in IT Sector, Healthcare, Aviation.

```python
#Product catagory
sns.set(rc= {'figure.figsize':(20,8)}) #(width, height)
ax=sns.countplot(x= 'Product_Category', data= df)
for bars in ax.containers:
 ax.bar_label(bars)
```

```python
Sales_product_cat = df.groupby(['Product_Category'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(25,6)})
sns.barplot(x= 'Product_Category', y= 'Amount', data= Sales_product_cat)
ax3 = sns.barplot(x= 'Product_Category', y= 'Amount', data= Sales_product_cat)
for bars in ax3.containers:
 ax3.bar_label(bars)
```

From above graphs we can conclude that, sold products are from Food, Clothing, and Electronics.

```python
#product ID
Sales_productID = df.groupby(['Product_ID'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False).head(10)
sns.set(rc= {'figure.figsize':(20,7)})
sns.barplot(x= 'Product_ID', y= 'Amount', data= Sales_productID)
ax3 = sns.barplot(x= 'Product_ID', y= 'Amount', data= Sales_productID)
for bars in ax3.containers:
 ax3.bar_label(bars)
```

```python
#top 10 most sold products (same thing as above)
fig1, ax1 = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending= False).plot(kind='bar')
for bars in ax1.containers:
 ax1.bar_label(bars)
```

### **CONCLUSION**
The Married women age group 26-35 Yearrs from UP, MH,KAR working in IT, Healthcare, Aviation are more likely to buy products from Food, Clothing, and Electronics Category.
